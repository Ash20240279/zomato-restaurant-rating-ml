    ▪️ 5.6 K-Fold Cross Validation Evaluation for Regression
  </h2>
  <p style="color: #94a3b8; font-size: 13.5px; margin: 0; line-height: 1.5;">
    <b>Quick Note:</b> Assessing model stability and generalization performance across 5 cross-validation folds.

```python
kf = KFold(n_splits=5, shuffle=True, random_state=42)
```

```python
regression_models = {
    'Linear Regression': lr_model,
    'KNN Regressor': knn_model,
    'Decision Tree': dt_model,
    'Random Forest': rf_model,
    'AdaBoost': ada_model,
    'Gradient Boosting': gbr_model
}
```

```python
cv_results = []
for name, model in regression_models.items():
    # Negative MSE to calculate RMSE
    mse_scores = cross_val_score(model, X_train_scaled, y_train, cv=kf, scoring='neg_mean_squared_error', n_jobs=-1)
    r2_scores = cross_val_score(model, X_train_scaled, y_train, cv=kf, scoring='r2', n_jobs=-1)
    
    rmse_scores = np.sqrt(-mse_scores)
    
    cv_results.append({
        'Model': name,
        'Mean CV RMSE': rmse_scores.mean(),
        'Std CV RMSE': rmse_scores.std(),
        'Mean CV R²': r2_scores.mean()
    })
```

```python
cv_df = pd.DataFrame(cv_results).sort_values(by='Mean CV R²', ascending=False)
display(cv_df)
```
